```python
import torch
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
from torchvision.datasets import ImageFolder
import matplotlib.pyplot as plt
```

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Redimensionner toutes les images à 224x224
    transforms.ToTensor(),  # Convertir en tenseur PyTorch
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])  # Normalisation pour ResNet
])

# Chargement du dataset
dataset = ImageFolder(root="content/sorted_movie_posters_paligema", transform=transform)

# DataLoader pour le train et le test
train_dataset, test_dataset = torch.utils.data.random_split(dataset, [int(0.8 * len(dataset)), len(dataset) - int(0.8 * len(dataset))])
dataloader_train = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=4)
dataloader_test = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=4)

# Affichage des classes détectées dans le train
print(train_dataset.dataset.classes)
print(train_dataset.dataset.class_to_idx)
print(test_dataset.dataset.classes)
```

```python
import torch.nn as nn
import torchvision.models as models

# Charger un modèle pré-entraîné
model = models.resnet18(pretrained=True)

# Adapter la dernière couche au nombre de classes détectées
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, len(dataset.classes))

# Déplacer le modèle sur GPU si disponible
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)
```

```python
import torch.optim as optim

# Définition de la fonction de perte et de l'optimiseur
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Boucle d'entraînement
num_epochs = 5

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0

    for images, labels in dataloader_train:
        images, labels = images.to(device), labels.to(device)
        print("label : ", labels[0])
        # plot une image
        plt.imshow(images[0].cpu().detach().permute(1, 2, 0))
        plt.show()

        # Forward
        outputs = model(images)
        loss = criterion(outputs, labels)

        # Backward
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    print(f"Époque {epoch+1}, Perte: {running_loss/len(dataloader_train)}")
```
